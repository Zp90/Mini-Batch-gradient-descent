```python
import numpy as np

X = np.arange(1,6)
Y = np.arange(1,6)

#Scale
def StandardScaling(array):
    mean = np.mean(array)
    SD = (np.sum((array - mean)**2) / len(array))**(1/2)
    array = (array - mean)/SD
    return array

X = StandardScaling(array=X)
Y= StandardScaling(array=Y)

X,Y
```

```python

O0 = 0
O1 = 0

# Graph =  O0*x + O1 = y
def LossFunction(O0,O1,x,y):
    return sum((y - O0*x - O1)**2) / len(x)

def DerivativeOfLostFunction(O0,O1,x,y):
    h = 0.001
    DerO0 = (LossFunction(O0+h,O1,x,y) - LossFunction(O0,O1,x,y))/h
    DerO1 = (LossFunction(O0,O1+h,x,y) - LossFunction(O0,O1,x,y))/h
    return DerO0,DerO1


BatchSize = 2
Epochs = 1
learningRate = 0.01

for i in range(Epochs):
    RandomPermutation = np.random.permutation(len(X))
    NewX = X[RandomPermutation]
    NewY = Y[RandomPermutation]

    for i2 in range(0,len(X),BatchSize):
        #Batch
        BatchX = NewX[i2:i2+BatchSize]
        BatchY = NewY[i2:i2+BatchSize]

        #learning
        DerO0,DerO1 = DerivativeOfLostFunction(O0,O1,BatchX,BatchY)
        O0 = O0 - learningRate*DerO0
        O1 = O1 - learningRate*DerO1

        break

print(O0,O1,LossFunction(O0,O1,X,Y))
```
